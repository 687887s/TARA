```python
import glob
import pandas as pd
df = pd.read_csv('../all_sample_sorted_by_o_with_classification(close companion).csv')
```

```python
tg_list = [f'../data/catalog/ps1/redshift/origin/t{ra:08.4f}{dec:+07.4f}.csv'for ra, dec in zip(df.paliya_RA, df.paliya_DEC)]
```

```python
import numpy as np
cat_list = np.array(glob.glob('../data/catalog/ps1/redshift/origin/*.csv'))
mask = np.array([i in tg_list for i in cat_list])
```

```python
cat_list = cat_list[mask]
```

```python
df_close = pd.DataFrame(columns = pd.read_csv(cat_list[0]).sort_values(by='distance').head(1)[['distance', 'raMean', 'decMean', 'z_phot', 'z_phot_err', 'Absolute_Mag_r']].columns)
for i in range(len(cat_list)):
    df_close.loc[i] = pd.read_csv(cat_list[i]).sort_values(by='distance').iloc[0][['distance', 'raMean', 'decMean', 'z_phot', 'z_phot_err', 'Absolute_Mag_r']]
```

```python
df = pd.concat([df, df_close], axis = 1)
```

```python
tg_list = [f'../data/catalog/ps1/redshift_xgb/origin/t{ra:08.4f}{dec:+07.4f}.csv'for ra, dec in zip(df.paliya_RA, df.paliya_DEC)]
```

```python
import numpy as np
cat_list = np.array(glob.glob('../data/catalog/ps1/redshift_xgb/origin/*.csv'))
mask = np.array([i in tg_list for i in cat_list])
```

```python
cat_list = cat_list[mask]
```

```python
df_close = pd.DataFrame(columns = ['z_phot_xgb', 'z_phot_err_xgb'])
for i in range(len(cat_list)):
    tmp = pd.read_csv(cat_list[i]).sort_values(by='distance').iloc[0][['z_phot', 'z_phot_err']]
    tmp.rename({'z_phot':'z_phot_xgb', 'z_phot_err':'z_phot_err_xgb'}, inplace = True)
    df_close.loc[i] = tmp
```

```python
df = pd.concat([df, df_close], axis=1)
```

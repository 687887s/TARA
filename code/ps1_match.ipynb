```python
import pandas as pd
import glob
```

```python
datas = sorted(glob.glob('../data/catalog/ps1/*.csv'))[400:]
z_phot = sorted(glob.glob('../data/catalog/ps1/redshift/origin/*.csv'))[400:]
z_phot_filtered = sorted(glob.glob('../data/catalog/ps1/redshift/filtered/*.csv'))[400:]
xgb_phot = sorted(glob.glob('../data/catalog/ps1/redshift_xgb/origin/*.csv'))[400:]
xgb_phot_filtered = sorted(glob.glob('../data/catalog/ps1/redshift_xgb/filtered/*.csv'))[400:]
```

```python
for target, z, z_filter, xgb_z, xgb_filter in zip(datas, z_phot, z_phot_filtered, xgb_phot, xgb_phot_filtered):
    df = pd.read_csv(target)
    df_z = pd.read_csv(z)
    df_z = df_z.drop(columns = df.columns[(df.columns == 'objID') == False])
    df_z_filter = pd.read_csv(z_filter)
    df_z_filter = df_z_filter.drop(columns = df.columns[(df.columns == 'objID') == False])
    df_xgb_z = pd.read_csv(xgb_z)
    # df_xgb_z = df_xgb_z.drop(columns = df.columns[(df.columns == 'objID') == False])
    df_xgb_filter = pd.read_csv(xgb_filter)
    # df_xgb_filter = df_xgb_filter(columns = df.columns[(df.columns == 'objID') == False])

    df = pd.merge(df, df_z, left_on='objID', right_on='objID', how = 'left', suffixes=('', '_llr_origin'))
    df = pd.merge(df, df_z_filter, left_on='objID', right_on='objID', how = 'left', suffixes=('', '_llr_filtered'))
    df = pd.merge(df, df_xgb_z, left_on='objID', right_on='objID', how = 'left', suffixes=('', '_xgb_origin'))
    df = pd.merge(df, df_xgb_filter, left_on='objID', right_on='objID', how = 'left', suffixes=('', '_xgb_filtered'))
    df.to_csv('../data/catalog/ps1/merged/' + target.split('../data/catalog/ps1/')[1])
```

```python
for target in datas:
    df = pd.read_csv(target)
    df = pd.merge(df, z_phot, left_on='objID', right_on='objID', how='left', suffixes=('', '_drop'))
    df = df.drop(columns = df.filter(regex = '_drop$').columns)
    df[pd.isna(df['z_phot']) == False].to_csv('../data/catalog/ps1/llr/result' +target.split('../data/catalog/ps1')[1])
```

```python
for target in datas:
    df = pd.read_csv(target)
    df = pd.merge(df, z_phot_filtered, left_on='objID', right_on='objID', how='left', suffixes=('', '_drop'))
    df = df.drop(columns = df.filter(regex = '_drop$').columns)
    df[pd.isna(df['z_phot']) == False].to_csv('../data/catalog/ps1/llr/result_filtered' +target.split('../data/catalog/ps1')[1])
```
